```python
# ============================================================
# CRUD OPERATIONS USING CHROMADB
# Create, Read, Update, Delete
# ============================================================

# Install required library
!pip install -q chromadb "opentelemetry-api<=1.42.1,>=1.39" "opentelemetry-sdk<=1.42.1,>=1.39"

import chromadb

# ------------------------------------------------------------
# 1. CREATE - Create ChromaDB client and collection
# ------------------------------------------------------------

client = chromadb.Client()

collection = client.get_or_create_collection(
    name="crud_demo"
)

print("Collection created successfully!")
print()


# ------------------------------------------------------------
# 2. CREATE - Add documents
# ------------------------------------------------------------

ids = ["doc1", "doc2", "doc3", "doc4", "doc5"]

documents = [
    "Machine learning is a branch of artificial intelligence.",
    "Python is a popular programming language.",
    "Deep learning uses neural networks.",
    "The weather is sunny today.",
    "Football is a popular sport."
]

metadatas = [
    {"category": "AI"},
    {"category": "Programming"},
    {"category": "AI"},
    {"category": "Weather"},
    {"category": "Sports"}
]

collection.add(
    ids=ids,
    documents=documents,
    metadatas=metadatas
)

print("CREATE:")
print("Documents added successfully.")
print("Number of documents:", collection.count())
print()


# ------------------------------------------------------------
# 3. READ - Read all documents
# ------------------------------------------------------------

print("READ - All Documents:")

all_data = collection.get()

for i in range(len(all_data["ids"])):
    print(
        all_data["ids"][i],
        "->",
        all_data["documents"][i],
        "->",
        all_data["metadatas"][i]
    )

print()


# ------------------------------------------------------------
# 4. READ - Read selected documents
# ------------------------------------------------------------

print("READ - Selected Documents:")

selected_data = collection.get(
    ids=["doc1", "doc3"]
)

for i in range(len(selected_data["ids"])):
    print(
        selected_data["ids"][i],
        "->",
        selected_data["documents"][i]
    )

print()


# ------------------------------------------------------------
# 5. READ - Semantic Search
# ------------------------------------------------------------

print("READ - Semantic Search:")

results = collection.query(
    query_texts=["artificial intelligence and neural networks"],
    n_results=3
)

for i in range(len(results["ids"][0])):
    print(
        results["ids"][0][i],
        "->",
        results["documents"][0][i]
    )

print()


# ------------------------------------------------------------
# 6. READ - Metadata Filtering
# ------------------------------------------------------------

print("READ - Metadata Filtering:")

ai_documents = collection.get(
    where={"category": "AI"}
)

for i in range(len(ai_documents["ids"])):
    print(
        ai_documents["ids"][i],
        "->",
        ai_documents["documents"][i]
    )

print()


# ------------------------------------------------------------
# 7. UPDATE - Update an existing document
# ------------------------------------------------------------

collection.update(
    ids=["doc2"],
    documents=[
        "Python is widely used for artificial intelligence and data science."
    ],
    metadatas=[
        {"category": "Programming", "updated": "yes"}
    ]
)

print("UPDATE:")
print("doc2 updated successfully.")

updated_data = collection.get(
    ids=["doc2"]
)

print(
    updated_data["ids"][0],
    "->",
    updated_data["documents"][0]
)

print()


# ------------------------------------------------------------
# 8. UPSERT - Add or update documents
# ------------------------------------------------------------

collection.upsert(
    ids=["doc5", "doc6"],
    documents=[
        "Football is played all around the world.",
        "Natural Language Processing works with human language."
    ],
    metadatas=[
        {"category": "Sports"},
        {"category": "NLP"}
    ]
)

print("UPSERT:")
print("doc5 updated and doc6 added.")
print("Number of documents:", collection.count())
print()


# ------------------------------------------------------------
# 9. DELETE - Delete one document
# ------------------------------------------------------------

collection.delete(
    ids=["doc4"]
)

print("DELETE:")
print("doc4 deleted successfully.")
print("Number of documents:", collection.count())
print()


# ------------------------------------------------------------
# 10. DELETE - Delete documents using metadata
# ------------------------------------------------------------

collection.delete(
    where={"category": "Sports"}
)

print("DELETE USING FILTER:")
print("All Sports documents deleted.")
print("Number of documents:", collection.count())
print()


# ------------------------------------------------------------
# 11. FINAL DATABASE
# ------------------------------------------------------------

print("FINAL DATABASE:")

final_data = collection.get()

for i in range(len(final_data["ids"])):
    print(
        final_data["ids"][i],
        "->",
        final_data["documents"][i],
        "->",
        final_data["metadatas"][i]
    )

print()


# ------------------------------------------------------------
# 12. CRUD SUMMARY
# ------------------------------------------------------------

print("=" * 50)
print("CRUD OPERATION SUMMARY")
print("=" * 50)

print("CREATE  -> Add documents using collection.add()")
print("READ    -> Retrieve documents using collection.get()")
print("SEARCH  -> Find similar documents using collection.query()")
print("UPDATE  -> Modify documents using collection.update()")
print("UPSERT  -> Add or update using collection.upsert()")
print("DELETE  -> Remove documents using collection.delete()")

print()
print("CRUD operations using ChromaDB completed successfully!")
```
