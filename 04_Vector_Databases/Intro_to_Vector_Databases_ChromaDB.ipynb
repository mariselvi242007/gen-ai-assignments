```python
# ============================================================
# INTRODUCTION TO VECTOR DATABASES USING CHROMADB
# ============================================================

# Install ChromaDB
!pip install -q chromadb "opentelemetry-api<=1.42.1,>=1.39" "opentelemetry-sdk<=1.42.1,>=1.39"

# Import library
import chromadb

print("=" * 60)
print("VECTOR DATABASE USING CHROMADB")
print("=" * 60)


# ------------------------------------------------------------
# 1. Create ChromaDB client
# ------------------------------------------------------------

client = chromadb.Client()

print("\nChromaDB client created successfully!")


# ------------------------------------------------------------
# 2. Create a collection
# ------------------------------------------------------------

collection = client.get_or_create_collection(
    name="my_documents"
)

print("\nCollection created:")
print(collection.name)


# ------------------------------------------------------------
# 3. Add documents
# ------------------------------------------------------------

documents = [
    "Machine learning is a branch of artificial intelligence.",
    "Deep learning uses neural networks.",
    "Python is widely used for data science.",
    "Natural language processing works with human language.",
    "Computer vision helps computers understand images."
]

ids = [
    "doc1",
    "doc2",
    "doc3",
    "doc4",
    "doc5"
]

metadatas = [
    {"category": "AI"},
    {"category": "AI"},
    {"category": "Programming"},
    {"category": "NLP"},
    {"category": "Computer Vision"}
]

collection.add(
    documents=documents,
    ids=ids,
    metadatas=metadatas
)

print("\nDocuments added successfully!")
print("Number of documents:", collection.count())


# ------------------------------------------------------------
# 4. Display stored documents
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("STORED DOCUMENTS")
print("-" * 60)

result = collection.get()

for doc_id, document, metadata in zip(
    result["ids"],
    result["documents"],
    result["metadatas"]
):
    print(f"\nID: {doc_id}")
    print(f"Document: {document}")
    print(f"Metadata: {metadata}")


# ------------------------------------------------------------
# 5. Peek at documents
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("PREVIEW OF DOCUMENTS")
print("-" * 60)

preview = collection.peek(limit=3)

for doc_id, document in zip(
    preview["ids"],
    preview["documents"]
):
    print(f"{doc_id} -> {document}")


# ------------------------------------------------------------
# 6. Get a specific document
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("GET SPECIFIC DOCUMENT")
print("-" * 60)

result = collection.get(
    ids=["doc1"]
)

print("ID:", result["ids"][0])
print("Document:", result["documents"][0])
print("Metadata:", result["metadatas"][0])


# ------------------------------------------------------------
# 7. Semantic search
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("SEMANTIC SEARCH")
print("-" * 60)

query = "Artificial intelligence and neural networks"

search_results = collection.query(
    query_texts=[query],
    n_results=3
)

print("Query:")
print(query)

print("\nMost relevant documents:")

for doc_id, document, distance in zip(
    search_results["ids"][0],
    search_results["documents"][0],
    search_results["distances"][0]
):
    print(
        f"\nID: {doc_id}"
        f"\nDistance: {distance:.4f}"
        f"\nDocument: {document}"
    )


# ------------------------------------------------------------
# 8. Search using metadata filter
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("FILTERED SEARCH")
print("-" * 60)

filtered_results = collection.query(
    query_texts=["artificial intelligence"],
    n_results=2,
    where={"category": "AI"}
)

print("Category filter: AI")

for doc_id, document in zip(
    filtered_results["ids"][0],
    filtered_results["documents"][0]
):
    print(f"\n{doc_id} -> {document}")


# ------------------------------------------------------------
# 9. Update a document
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("UPDATE DOCUMENT")
print("-" * 60)

collection.update(
    ids=["doc3"],
    documents=[
        "Python is widely used for data science, machine learning and AI."
    ],
    metadatas=[
        {"category": "Programming", "updated": True}
    ]
)

updated = collection.get(
    ids=["doc3"]
)

print("Updated document:")
print(updated["documents"][0])

print("\nUpdated metadata:")
print(updated["metadatas"][0])


# ------------------------------------------------------------
# 10. Upsert a document
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("UPSERT DOCUMENT")
print("-" * 60)

collection.upsert(
    ids=["doc6"],
    documents=[
        "Generative AI can create text, images and other content."
    ],
    metadatas=[
        {"category": "Generative AI"}
    ]
)

print("New document added using upsert.")

print("\nTotal documents:",
      collection.count())


# ------------------------------------------------------------
# 11. Delete a document
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("DELETE DOCUMENT")
print("-" * 60)

collection.delete(
    ids=["doc5"]
)

print("doc5 deleted.")

print("Remaining document count:",
      collection.count())


# ------------------------------------------------------------
# 12. Display final database
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("FINAL DATABASE")
print("-" * 60)

final_result = collection.get()

for doc_id, document, metadata in zip(
    final_result["ids"],
    final_result["documents"],
    final_result["metadatas"]
):
    print(f"\n{doc_id}")
    print(document)
    print(metadata)


# ------------------------------------------------------------
# 13. Final result
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("CHROMADB VECTOR DATABASE COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nCollection:", collection.name)
print("Final document count:", collection.count())

print("\nOperations demonstrated:")
print("1. Create ChromaDB client")
print("2. Create collection")
print("3. Add documents")
print("4. Store metadata")
print("5. Get documents")
print("6. Semantic search")
print("7. Metadata filtering")
print("8. Update documents")
print("9. Upsert documents")
print("10. Delete documents")

print("\nVector database demonstration completed!")
```
