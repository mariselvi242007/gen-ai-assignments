```python
# ============================================================
# WORD EMBEDDINGS USING WORD2VEC
# ============================================================

# Install required library
!pip install -q gensim

# Import libraries
from gensim.models import Word2Vec
import numpy as np


# ------------------------------------------------------------
# 1. Create sample sentences
# ------------------------------------------------------------

sentences = [
    ["machine", "learning", "is", "interesting"],
    ["machine", "learning", "uses", "data"],
    ["deep", "learning", "uses", "neural", "networks"],
    ["neural", "networks", "are", "powerful"],
    ["python", "is", "used", "for", "machine", "learning"],
    ["data", "science", "uses", "machine", "learning"],
    ["artificial", "intelligence", "is", "growing"],
    ["machine", "learning", "is", "part", "of",
     "artificial", "intelligence"]
]

print("=" * 60)
print("WORD EMBEDDINGS USING WORD2VEC")
print("=" * 60)

print("\nNumber of sentences:", len(sentences))


# ------------------------------------------------------------
# 2. Train Word2Vec model
# ------------------------------------------------------------

model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1
)

print("\nWord2Vec model created successfully!")


# ------------------------------------------------------------
# 3. Display vocabulary
# ------------------------------------------------------------

vocabulary = list(model.wv.index_to_key)

print("\nVocabulary:")
print(vocabulary)

print("\nVocabulary size:", len(vocabulary))


# ------------------------------------------------------------
# 4. Get embedding for a word
# ------------------------------------------------------------

word = "machine"

vector = model.wv[word]

print("\n" + "-" * 60)
print("WORD EMBEDDING")
print("-" * 60)

print("Word:", word)
print("Embedding vector:")
print(vector)

print("\nEmbedding dimension:", len(vector))


# ------------------------------------------------------------
# 5. Find similar words
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("SIMILAR WORDS")
print("-" * 60)

word = "learning"

similar_words = model.wv.most_similar(word, topn=5)

print("Words similar to:", word)

for similar_word, similarity in similar_words:
    print(f"{similar_word}: {similarity:.4f}")


# ------------------------------------------------------------
# 6. Compare two words
# ------------------------------------------------------------

word1 = "machine"
word2 = "learning"

similarity = model.wv.similarity(word1, word2)

print("\n" + "-" * 60)
print("WORD SIMILARITY")
print("-" * 60)

print(f"Similarity between '{word1}' and '{word2}': {similarity:.4f}")


# ------------------------------------------------------------
# 7. Calculate cosine similarity manually
# ------------------------------------------------------------

vector1 = model.wv[word1]
vector2 = model.wv[word2]

cosine_similarity = np.dot(vector1, vector2) / (
    np.linalg.norm(vector1) * np.linalg.norm(vector2)
)

print("\n" + "-" * 60)
print("MANUAL COSINE SIMILARITY")
print("-" * 60)

print(
    f"Cosine similarity between '{word1}' and '{word2}': "
    f"{cosine_similarity:.4f}"
)


# ------------------------------------------------------------
# 8. Compare another pair
# ------------------------------------------------------------

word1 = "neural"
word2 = "networks"

similarity = model.wv.similarity(word1, word2)

print("\nSimilarity between "
      f"'{word1}' and '{word2}': {similarity:.4f}")


# ------------------------------------------------------------
# 9. Display selected word embeddings
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("SELECTED WORD EMBEDDINGS")
print("-" * 60)

words = ["machine", "learning", "data", "python", "neural"]

for word in words:
    print(f"\n{word}:")
    print(model.wv[word])


# ------------------------------------------------------------
# 10. Save the Word2Vec model
# ------------------------------------------------------------

model.save("word_embeddings.model")

print("\n" + "-" * 60)
print("MODEL SAVED")
print("-" * 60)

print("Word2Vec model saved as: word_embeddings.model")


# ------------------------------------------------------------
# 11. Load the saved model
# ------------------------------------------------------------

loaded_model = Word2Vec.load("word_embeddings.model")

print("\nModel loaded successfully!")


# ------------------------------------------------------------
# 12. Final result
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("WORD EMBEDDINGS COMPLETED SUCCESSFULLY")
print("=" * 60)

print("Vocabulary size:", len(model.wv.index_to_key))
print("Embedding dimension:", model.vector_size)
print("Example word:", "machine")

print("\nTop similar words to 'machine':")

for word, score in model.wv.most_similar("machine", topn=5):
    print(f"{word} -> {score:.4f}")
```
