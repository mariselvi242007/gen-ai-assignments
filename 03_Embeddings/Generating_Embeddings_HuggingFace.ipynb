```python
# ============================================================
# GENERATING TEXT EMBEDDINGS USING HUGGING FACE
# ============================================================

# Install required libraries
!pip install -q sentence-transformers

# Import library
from sentence_transformers import SentenceTransformer
import numpy as np

print("=" * 60)
print("TEXT EMBEDDINGS USING HUGGING FACE")
print("=" * 60)


# ------------------------------------------------------------
# 1. Load pre-trained Hugging Face model
# ------------------------------------------------------------

model_name = "sentence-transformers/all-MiniLM-L6-v2"

model = SentenceTransformer(model_name)

print("\nModel loaded successfully!")
print("Model:", model_name)


# ------------------------------------------------------------
# 2. Create sample sentences
# ------------------------------------------------------------

sentences = [
    "Machine learning is a branch of artificial intelligence.",
    "Deep learning uses neural networks.",
    "Python is a popular programming language.",
    "I enjoy playing football.",
    "The weather is sunny today."
]

print("\n" + "-" * 60)
print("INPUT SENTENCES")
print("-" * 60)

for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence}")


# ------------------------------------------------------------
# 3. Generate embeddings
# ------------------------------------------------------------

embeddings = model.encode(sentences)

print("\n" + "-" * 60)
print("GENERATED EMBEDDINGS")
print("-" * 60)

print("Embedding shape:", embeddings.shape)

print("\nEach sentence has an embedding vector of size:",
      embeddings.shape[1])


# ------------------------------------------------------------
# 4. Display embedding of first sentence
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("EMBEDDING OF FIRST SENTENCE")
print("-" * 60)

print("Sentence:")
print(sentences[0])

print("\nEmbedding vector:")
print(embeddings[0])


# ------------------------------------------------------------
# 5. Display first 10 values of each embedding
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("FIRST 10 VALUES OF EACH EMBEDDING")
print("-" * 60)

for i, embedding in enumerate(embeddings):
    print(f"\nSentence {i + 1}:")
    print(embedding[:10])


# ------------------------------------------------------------
# 6. Calculate cosine similarity
# ------------------------------------------------------------

def cosine_similarity(vector1, vector2):
    return np.dot(vector1, vector2) / (
        np.linalg.norm(vector1) * np.linalg.norm(vector2)
    )


# ------------------------------------------------------------
# 7. Compare first two sentences
# ------------------------------------------------------------

similarity = cosine_similarity(
    embeddings[0],
    embeddings[1]
)

print("\n" + "-" * 60)
print("COSINE SIMILARITY")
print("-" * 60)

print("Sentence 1:")
print(sentences[0])

print("\nSentence 2:")
print(sentences[1])

print("\nCosine similarity:", round(float(similarity), 4))


# ------------------------------------------------------------
# 8. Compare all sentences
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("PAIRWISE SENTENCE SIMILARITY")
print("-" * 60)

for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):

        similarity = cosine_similarity(
            embeddings[i],
            embeddings[j]
        )

        print(
            f"Sentence {i + 1} vs Sentence {j + 1}: "
            f"{similarity:.4f}"
        )


# ------------------------------------------------------------
# 9. Find most similar sentence to a query
# ------------------------------------------------------------

query = "Artificial intelligence and machine learning"

query_embedding = model.encode([query])[0]

print("\n" + "-" * 60)
print("SEMANTIC SEARCH")
print("-" * 60)

print("Query:")
print(query)

similarities = []

for i, embedding in enumerate(embeddings):
    similarity = cosine_similarity(
        query_embedding,
        embedding
    )

    similarities.append((i, similarity))


# Sort by similarity
similarities.sort(
    key=lambda x: x[1],
    reverse=True
)


print("\nMost similar sentences:")

for i, similarity in similarities:
    print(
        f"{similarity:.4f} -> {sentences[i]}"
    )


# ------------------------------------------------------------
# 10. Generate embedding for a new sentence
# ------------------------------------------------------------

new_sentence = "Neural networks are used in artificial intelligence."

new_embedding = model.encode([new_sentence])[0]

print("\n" + "-" * 60)
print("NEW SENTENCE EMBEDDING")
print("-" * 60)

print("Sentence:")
print(new_sentence)

print("\nEmbedding dimension:")
print(len(new_embedding))

print("\nFirst 10 values:")
print(new_embedding[:10])


# ------------------------------------------------------------
# 11. Compare new sentence with existing sentences
# ------------------------------------------------------------

print("\n" + "-" * 60)
print("NEW SENTENCE SIMILARITY")
print("-" * 60)

results = []

for i, embedding in enumerate(embeddings):

    similarity = cosine_similarity(
        new_embedding,
        embedding
    )

    results.append((i, similarity))


results.sort(
    key=lambda x: x[1],
    reverse=True
)


for i, similarity in results:
    print(
        f"{similarity:.4f} -> {sentences[i]}"
    )


# ------------------------------------------------------------
# 12. Save embeddings
# ------------------------------------------------------------

np.save("sentence_embeddings.npy", embeddings)

print("\n" + "-" * 60)
print("EMBEDDINGS SAVED")
print("-" * 60)

print("Saved as: sentence_embeddings.npy")


# ------------------------------------------------------------
# 13. Load embeddings
# ------------------------------------------------------------

loaded_embeddings = np.load(
    "sentence_embeddings.npy"
)

print("\nEmbeddings loaded successfully!")

print("Loaded embedding shape:",
      loaded_embeddings.shape)


# ------------------------------------------------------------
# 14. Final result
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TEXT EMBEDDING GENERATION COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nModel:", model_name)

print("Number of sentences:",
      len(sentences))

print("Embedding dimension:",
      embeddings.shape[1])

print("\nEmbeddings were generated using Hugging Face.")
```
